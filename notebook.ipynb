```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import AzureChatOpenAI

from dotenv import load_dotenv
%load_ext autoreload
%autoreload 2

load_dotenv(dotenv_path="internal/.env") 

llm = AzureChatOpenAI(
    model="gpt-4o-mini", 
    temperature=0,
    max_tokens=None,
    timeout=None,
    max_retries=2,

)
```

```python
import os
import sys
sys.path.insert(0, os.getcwd())

from prompts.kg import PROMPTS
print(PROMPTS.keys())
```

```python
import pandas as pd
df = pd.read_csv("input/Ensemble_learning.csv")
texts = df['content'].tolist()
print("Chunk snippet (first 100 char):\n\n",texts[:1][0][:100])
```

```python
prompt = ChatPromptTemplate.from_template(PROMPTS['build_kg'])

chain = prompt | llm | StrOutputParser()
```

```python
response = chain.invoke({"input_text": texts[2]})
```

```python
import pandas as pd
import random
random.seed(42)

df = pd.read_csv("input/Ensemble_learning.csv")
texts = df['content'].tolist()

SUBSET_SIZE = 200

texts = random.sample(texts, SUBSET_SIZE)
print("Chunk snippet (first 100 char):\n\n",texts[:1][0][:100])

# join all texts into a single string
text = "\n".join(texts)
# save the text to a file
with open(f"input/Ensemble_learning_sub_{SUBSET_SIZE}.txt", "w") as f:
    f.write(text)
```
